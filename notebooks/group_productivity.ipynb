```python
import pandas as pd
import sys

from pathlib import Path

REPO_ROOT = Path.cwd().parent
# make the shared code in src/ importable even when the package is not pip-installed
if str(REPO_ROOT / "src") not in sys.path:
    sys.path.insert(0, str(REPO_ROOT / "src"))
```

```python
include_salinity = True
ribasim_run = 50
```

```python
base_excel_path = "/Users/hemert/OneDrive - Stichting Deltares/Tiaravanni Hermawan's files - Egypt_ERF_data/data_correlation.xlsx"
if include_salinity:
    new_excel_path = Path(base_excel_path).with_name(f'data_correlation_{ribasim_run}.xlsx')
else:
    new_excel_path = Path(base_excel_path).with_name(f'data_correlation_{ribasim_run}_no_salinity.xlsx')
```

```python
result_dict = {}
for ribasim_run in [30, 40, 50, 60, 70, 80, 90, 95]:
    excel_file_salinity = Path(base_excel_path).with_name(f'data_correlation_{ribasim_run}.xlsx')
    excel_file_no_sailinity = Path(base_excel_path).with_name(f'data_correlation_{ribasim_run}_no_salinity.xlsx')

    salinity_df = pd.read_excel(excel_file_salinity, sheet_name="command_unit")
    no_salinity_df = pd.read_excel(excel_file_no_sailinity, sheet_name="command_unit")

    result_dict[f'agricultural_productivity_{ribasim_run}'] = salinity_df['V']
    result_dict[f'water_procuctivity_{ribasim_run}'] = salinity_df['W']

    result_dict[f'agricultural_productivity_no_salinity_{ribasim_run}'] = no_salinity_df['V']
    result_dict[f'water_procuctivity_no_salinity_{ribasim_run}'] = no_salinity_df['W']

result_df = pd.DataFrame(result_dict)
result_df['area'] = salinity_df['area']
result_df.to_csv(REPO_ROOT / 'output/group_productivity.csv')
```
